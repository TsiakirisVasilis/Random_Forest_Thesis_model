```python
import pandas as pd
import numpy as np
import os
import glob
```

#### Preprocess the final excel files

```python
csvs = [r"/content/drive/MyDrive/EarthEngine_Exports/Final_complete_NO_SM_dataset.csv"]

# Get all CSV files in the folder
df = pd.concat((pd.read_csv(file) for file in csvs), ignore_index=True)
```

##### First get rid of the Null values

```python
clean_df = df.dropna()
```

##### Second rescale and convert values

```python
clean_df.loc[:, 'NDVI'] = clean_df['NDVI'] * 0.0001
clean_df.loc[:, 'ET'] = clean_df['ET'] * 0.1
clean_df.loc[:, 'LST_Day_1km'] = clean_df['LST_Day_1km']  * 0.02 - 273.15
clean_df.loc[:, 'land_cover'] = clean_df['land_cover'].round(0)
clean_df.head(5)
```

##### One hot encoding for the land cover category

```python
landcover_map = {
10 : "Tree cover",
20 : "Shrubland",
30 : "Grassland",
40 : "Cropland",
50 : "Built-up",
60 : "Bare / sparse vegetation",
70 : "Snow and ice",
80 : "Permanent water bodies",
90 : "Herbaceous wetland",
95 : "Mangroves",
100 : "Moss and lichen"
}
clean_df.loc[:, 'LandCover_label'] = clean_df['land_cover'].map(landcover_map)

clean_df = pd.get_dummies(clean_df, columns=['LandCover_label'], prefix='', prefix_sep='')
```

```python
final_df = clean_df.drop(columns=['date', 'dgg_id', 'Date', 'land_cover'])
final_df.to_csv(r"/content/drive/MyDrive/EarthEngine_Exports\Final_Combined_Processed_NO_SM.csv")
```
