```python
# Load the data first
data = pd.read_csv(r"/content/drive/MyDrive/EarthEngine_Exports/Final_Combined_Processed_NO_SM.csv")
data.columns
data["Soil_Moisture"] = np.nan
X = data.drop(["Soil_Moisture_SMAP", "Unnamed: 0", "Latitude", "Longitude"], axis=1)   # predictors
y = data["Soil_Moisture"]
```

```python
# Load the saved model
model = joblib.load(r"C:\Users\Bill Tsiakiris\Downloads\random_forest_model_KFold.pkl")

predictions = model.predict(X) # Make predictions
data["Soil_Moisture"] = predictions

# Export to csv
data.to_csv(r"/content/drive/MyDrive/EarthEngine_Exports/Final_predictions.csv", index=False)
```

```python
# Evaluation Metrics

# True values (SMAP)
y_true = data["Soil_Moisture_SMAP"]

rmse = np.sqrt(mean_squared_error(y_true, predictions))
mae = mean_absolute_error(y_true, predictions)
r2 = r2_score(y_true, predictions)
bias = np.mean(predictions - y_true)
corr = np.corrcoef(y_true, predictions)[0,1]

print("RMSE :", rmse)
print("MAE  :", mae)
print("R²   :", r2)
print("Bias :", bias)
print("Corr :", corr)
```
